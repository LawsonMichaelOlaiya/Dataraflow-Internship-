````python
# 1. **Student Grades System**  
#    Write a program to accept marks for 5 subjects, calculate the average, and display the grade based on:  
#    - A: 80–100  
#    - B: 60–79  
#    - C: 40–59  
#    - F: < 40  


marks = []

for i in range(5):
    mark = float(input(f"Enter marks for subject {i + 1}: "))
    marks.append(mark)

average = sum(marks) / 5

if average >= 80:
    grade = "A"
elif average >= 60:
    grade = "B"
elif average >= 40:
    grade = "C"
else:
    grade = "D"


print(f"Average is {average}, Grade is {grade}")
````

````python
# 2. **Simple Calculator**  
#    Build a calculator program that can perform addition, subtraction, multiplication, and division based on user input.


num1 = float(input("Enter the first number"))
operator = input("Enter an operator (+, -, *, /)")
num2 = float(input("Enter second number"))

if operator == "+":
    result = num1 + num2
elif operator == "-":
    result = num1 - num2
elif operator == "*":
    result = num1 * num2
elif operator == "/":
    if num2 != 0:
        result = num1 / num2
    else:
        result = "Error: cannot divide by zero"
else:
    result = "Invalid Operator"


print(f"{num1} {operator} {num2} equals {result}")
````

````python
# 3. **Shopping Cart**  
#    Create a program where a user can add items to a shopping cart (dictionary with item names and prices), then display the total cost.

cart = {}

while True:
    item = input("Enter item name or type DONE to finish")
    if item.lower() == "done":
        break

    price = float(input("Enter item price"))
    cart[item] = price

total = sum(cart.values())

print("\nShopping Cart:")

for item, price in cart.items():
    print(f"{item} #{price}")

print(f"Total cost: #{total}")
````

````python
# 4. **Guessing Game**  
#    Write a program where the computer randomly selects a number between 1–20, and the user must guess it. Give hints if the guess is too high or too low.

import random

number = random.randint(1,20)

while True:
    guess = int(input("Guess a number between 1 and 20"))

    if guess < number:
        print("Too low! Try Again")

    elif guess > number:
        print("Too high! Try again")

    else:
        print("Correct! You guessed the number!")
        break
````

````python
# 5. **File Handling**  
#    Write a program that accepts a list of student names from the user, saves them in a text file, and then reads and displays them.

students = []

number = int(input("How many students?"))

for i in range(number):
    name = input(f"Enter student {i + 1} name: ")
    students.append(name)

with open("students.txt", "w") as file:
    for student in students:
        file.write(student + "\n")

with open("students.txt", "r") as file:
    names = file.readlines()

print("\nStudents in the file:")

for name in names:
    print(name.strip())
````

## 14. Write a program to check if a string is a palindrome.  


````python
# 14. Write a program to check if a string is a palindrome.  

string = input("Enter a word")

if string == string[::-1]:
    print(f"The word {string} is a palindrome!")
else:
    print(f"The word {string} is not a palindrome!")
````
