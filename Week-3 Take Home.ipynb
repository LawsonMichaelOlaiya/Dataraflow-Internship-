```python
# 1. **JSON File Handling**  
#    Write a program that stores a list of student records (name, age, grade) into a JSON file, and then reads the file back into a Python object.  

import json
records = [{"name":"Michael", "age":21, "grade":"A+"},
           {"name": "John", "age": 20, "grade": "A"},
           {"name": "Mary", "age": 21, "grade": "B"},
           {"name": "Peter", "age": 19, "grade": "A"}]


with open("students.json", "w") as file:
    json.dump(records, file, indent=4)

with open("students.json", "r") as file:
    students_from_file = json.load(file)

print(students_from_file)
```

```python
# 2. **Math Quiz**  
#    Build a program that asks the user 5 random math questions (addition, subtraction, multiplication, division) and keeps score of correct answers.  

import random

score = 0

for i in range(5):

    num1 = random.randint(1, 10)
    num2 = random.randint(1, 10)

    operation = random.choice(["+", "-", "*", "/"])

    if operation == "+":
        answer = num1 + num2

    elif operation == "-":
        answer = num1 - num2

    elif operation == "*":
        answer = num1 * num2

    else:
        answer = num1 / num2

    print(f"What is {num1} {operation} {num2}?")

    user_answer = float(input(f"What is {num1} {operation} {num2}?"))

    if user_answer == answer:
        print("Correct!")
        score += 1
    else:
        print(f"Wrong! The answer is {answer}")

print(f"You scored {score}/5")
```

```python
# 3. **Error Handling System**  
#    Create a program that asks the user for a number and divides 100 by that number. Use `try/except` to handle invalid input and division by zero.  

try:
    num = int(input("Enter a number"))
    divide = 100/num
    print(divide)
except ZeroDivisionError:
    print("Cannot divide by zero")
except:
    print("Invalid input")
```

```python
# 4. **Date Calculator**  
#    Write a program that takes a user’s birthday as input and tells them how many days are left until their next birthday.  

from datetime import datetime, date

bday = input("Enter your DOB (YYYY-MM-DD): ")

birthday = datetime.strptime(bday, "%Y-%m-%d").date()

today = date.today()

next_birthday = birthday.replace(year=today.year)

if next_birthday < today:
    next_birthday = birthday.replace(year=today.year + 1)

days_left = (next_birthday - today).days

print(f"There are {days_left} days left until your next birthday!")
```
