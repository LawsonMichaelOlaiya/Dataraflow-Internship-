```python
# 5. Use the `math` module to calculate the factorial of 6.    

import math

factorial = math.factorial(6)

print(factorial)
```

```python
# 6. Write a program that takes a user’s birth year as input and calculates their age.    

import datetime


while True:
    num = input("Enter your birth year....")
    try:
        year = int(num)
        today = int(datetime.datetime.now().strftime("%Y"))
        age = today - year
        print(f"birth year is {year}, current year: {today} \nage is {age}")
        break
    except:
        print("Enter a year")
```
